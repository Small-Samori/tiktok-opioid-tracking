```python
import pandas as pd
from sqlalchemy import create_engine
from tqdm import tqdm
import os
wkdir = os.getcwd()
mysql_engine = create_engine(
    "mysql+pymysql://iasamori@localhost/iasamori?charset=utf8mb4"
)
# SQLite engine (creates a .db file)
sqlite_engine = create_engine(f"sqlite:///{wkdir}/data/tiktok_dlatk_data.db")

tables_to_export = [
    "feat$1gram$cmt7_cln$message_id",
    "feat$1gram$cmt7_cln$message_id$0_0001",
    "feat$1gram$cmt7_cln$month_year",
    "feat$cat_LIWC2015_w$cmt7_cln$message_id$1gra",
    "feat$cat_LIWC2015_w$cmt7_cln$month_year$1gra",
    "feat$cat_tpc7_200_a2s100_cp_w$cmt7_cln$message_id$1gra",
    "feat$cat_tpc7_200_a2s100_cp_w$cmt7_cln$month_year$1gra",
    "feat$meta_1gram$cmt7_cln$message_id",
    "feat$meta_1gram$cmt7_cln$month_year",
    "lwic_month_year",
    "meta_tcp7_my",
    "tpc7_200_a2s100_cp",
    "tpc7_200_a2s100_freq_t50ll"]


for table in tqdm(tables_to_export):
    df = pd.read_sql(f"SELECT * FROM {table}", mysql_engine)
    df.to_sql(table, sqlite_engine, if_exists="replace", index=False)
```
